```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
```

```python
drop_cols = ["Attrition", "EmployeeCount", "Over18", "StandardHours", "EmployeeNumber"]
X = df.drop(columns=drop_cols)
```

```python
y = (df["Attrition"] == "Yes").astype(int).values
```

```python
cat_cols = X.select_dtypes(include="object").columns.tolist()
num_cols = [c for c in X.columns if c not in cat_cols]
```

```python
cat_cols
```

```python
X = pd.get_dummies(X, columns=cat_cols, drop_first=True).astype("float32")
X.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_tr, X_tmp, y_tr, y_tmp = train_test_split(X, y, test_size=0.30, stratify=y, random_state=SEED)
X_va, X_te, y_va, y_te = train_test_split(X_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=SEED)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score, average_precision_score, confusion_matrix, precision_recall_curve
```

```python
scaler = StandardScaler().fit(X_tr[num_cols])
for d in (X_tr, X_va, X_te):
    d[num_cols] = scaler.transform(d[num_cols])
    
to3d = lambda d: d.values.reshape(-1, 1, d.shape[1])
X_tr3, X_va3, X_te3 = map(to3d, (X_tr, X_va, X_te))
print(X_tr3.shape, X_va3.shape, X_te3.shape)
```

```python
model = keras.Sequential([
    layers.Input(shape=(1, X_tr3.shape[2])),
    layers.LSTM(32),
    layers.Dropout(0.3),
    layers.Dense(16, activation="relu"),
    layers.Dense(1, activation="sigmoid"),
])
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[keras.metrics.AUC(name="auc"),
             keras.metrics.AUC(name="pr_auc", curve="PR"),
             keras.metrics.Recall(name="recall")],
)
model.summary()
```

```python
history = model.fit(
    X_tr3, y_tr,
    validation_data=(X_va3, y_va),
    epochs=80, batch_size=32,
    class_weight={0: 1.0, 1: 5.0},
    callbacks=[keras.callbacks.EarlyStopping(monitor="val_pr_auc", mode="max",
                                             patience=10, restore_best_weights=True)],
    verbose=2,
)
```

```python
p_va = model.predict(X_va3).ravel()
prec, rec, thr = precision_recall_curve(y_va, p_va)
f1 = 2 * prec * rec / (prec + rec + 1e-9)
best_thr = thr[np.argmax(f1[:-1])]
print("best threshold (val F1):", round(float(best_thr), 3))

p_te = model.predict(X_te3).ravel()
pred = (p_te >= best_thr).astype(int)

print("ROC-AUC:", roc_auc_score(y_te, p_te))
print("PR-AUC :", average_precision_score(y_te, p_te))
print(confusion_matrix(y_te, pred))
print(classification_report(y_te, pred, target_names=["Stay", "Leave"]))
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=2000, class_weight="balanced").fit(X_tr, y_tr)
p_lr = lr.predict_proba(X_te)[:, 1]
print("LogReg ROC-AUC:", roc_auc_score(y_te, p_lr), "| PR-AUC:", average_precision_score(y_te, p_lr))
print("LSTM   ROC-AUC:", roc_auc_score(y_te, p_te), "| PR-AUC:", average_precision_score(y_te, p_te))
```
